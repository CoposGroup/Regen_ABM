```python
import numpy as np

# Create figure with larger size
plt.figure(figsize=(12, 6))

# Set width of bars and positions of the bars
bar_width = 0.35
x = np.arange(len(lengths))

# Create bars
bars1 = plt.bar(x - bar_width/2, list(lengths.values()), bar_width, 
                label='Hard Boundary', color='red', alpha=0.7)
bars2 = plt.bar(x + bar_width/2, [lengths_soft[k] for k in lengths.keys()], 
                bar_width, label='Soft Boundary', color='pink', alpha=0.7)

# Customize plot
plt.title('Growth Length Comparison: Hard vs Soft Boundary', fontsize=14)
plt.xlabel('Simulation Number', fontsize=12)
plt.ylabel('Growth Length', fontsize=12)

# Add value labels on top of each bar
def autolabel(bars):
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.2f}',
                ha='center', va='bottom')

autolabel(bars1)
autolabel(bars2)

# Set x-axis ticks
plt.xticks(x, lengths.keys())

# Add grid and legend
plt.grid(True, alpha=0.3, axis='y')
plt.legend()

# Adjust y-axis to start from 0
plt.ylim(0, max(max(lengths.values()), max(lengths_soft.values())) * 1.1)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

# Create figure with larger size
plt.figure(figsize=(14, 6))

# Set width of bars and positions of the bars
bar_width = 0.35

# Add '6_1' and '6_2' to the x-axis positions
all_labels = list(lengths.keys()) + ['6_1', '6_2']
x = np.arange(len(all_labels))

# Create bars
bars1 = plt.bar(x - bar_width/2, 
                [lengths.get(k, np.nan) for k in all_labels], 
                bar_width, 
                label='Hard Boundary', 
                color='red', 
                alpha=0.7)

# Get soft boundary values, including 6_1 and 6_2
soft_values = []
for k in all_labels:
    if k in lengths_soft:
        soft_values.append(lengths_soft[k])
    elif k == '6_1':
        soft_values.append(lengths_soft['6_1'])
    elif k == '6_2':
        soft_values.append(lengths_soft['6_2'])
    else:
        soft_values.append(np.nan)

bars2 = plt.bar(x + bar_width/2, 
                soft_values, 
                bar_width, 
                label='Soft Boundary', 
                color='pink', 
                alpha=0.7)

# Customize plot
plt.title('Growth Length Comparison: Hard vs Soft Boundary', fontsize=14)
plt.xlabel('Simulation Number', fontsize=12)
plt.ylabel('Growth Length', fontsize=12)

# Add value labels on top of each bar
def autolabel(bars):
    for bar in bars:
        height = bar.get_height()
        if not np.isnan(height):  # Only label non-NaN values
            plt.text(bar.get_x() + bar.get_width()/2., height,
                    f'{height:.2f}',
                    ha='center', va='bottom')

autolabel(bars1)
autolabel(bars2)

# Set x-axis ticks
plt.xticks(x, all_labels)

# Add grid and legend
plt.grid(True, alpha=0.3, axis='y')
plt.legend()

# Adjust y-axis to start from 0
plt.ylim(0, max(max(lengths.values()), max(lengths_soft.values())) * 1.1)

plt.tight_layout()
plt.show()
```
